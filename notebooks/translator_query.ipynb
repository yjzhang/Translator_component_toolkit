# Building and executing Translator queries using TCT

This notebook provides examples of the usage of the `TCT.translator_query` module.

```python
from TCT import name_resolver, translator_query
```

We want to answer questions of the form, what is a path between node A and node B?

```python
name_resolver.lookup?
```

```python
results = name_resolver.lookup('IFNG', return_top_response=False, biolink_type='biolink:Gene')
node1 = [r for r in results if 'NCBITaxon:9606' in r.taxa][0]
```

```python
node2 = name_resolver.lookup('COVID-19', return_top_response=True, biolink_type='biolink:Disease')
```

```python
node1
```

```python
node2
```

## Getting available Translator APIs

This is done using the function `translator_query.get_translator_API_predicates`

```python
APInames, metaKG, API_predicates = translator_query.get_translator_API_predicates()
```

```python
APInames
```

```python
metaKG
```

```python
API_predicates
```

## Building a query

Question that we want to answer: which drugs treat COVID-19?

For a list of available biolink types and predicates, see https://biolink.github.io/biolink-model/

In this case, we want to find all `biolink:Gene` entities that are linked to the IFNG node via the predicate `biolink:treats`.

```python
query = translator_query.build_query_json([node1.curie], ['biolink:Gene'], ['biolink:interacts_with'])
```

```python
query
```

## Selecting the APIs

Which APIs have the relations that we are looking for?

```python
from TCT import TCT
```

```python
selected_predicates, selected_APIs, API_URLs = TCT.select_predicates_API(['biolink:Gene'], ['biolink:Gene'], metaKG, APInames)
```

```python
selected_APIs
```

```python
selected_predicates
```

## Executing the query



```python
query_result = translator_query.parallel_api_query(query, selected_APIs, APInames, API_predicates)
```
